# Check ASP trends

```python
# Import Packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
data2025=pd.read_csv('2025.csv',encoding='utf-8')
data2026=pd.read_csv('2026PartYear.csv',encoding='utf-8')
```

```python
data2025.columns
data2026.columns
```

```python
data2025['year']=2025
data2026['year']=2026
```

```python
asp_2025=data2025.groupby('Geography')['ASP Current Year'].mean()
asp_2026=data2026.groupby('Geography')['ASP Current Year'].mean()

asp_compare = pd.DataFrame({"ASP_2025": asp_2025,"ASP_2026": asp_2026})
asp_compare["YOY_%_Change"] = ((asp_compare["ASP_2026"]-asp_compare["ASP_2025"]) / asp_compare["ASP_2025"]*100)

asp_compare
```

```python
vol_2025 = data2025.groupby("Geography")["Total Bulk and Bags"].sum()
vol_2026 = data2026.groupby("Geography")["Total Bulk and Bags"].sum()

vol_compare = pd.DataFrame({"Volume_2025": vol_2025,"Volume_2026": vol_2026})
vol_compare["YOY_%_Change"]=((vol_compare["Volume_2026"]-vol_compare["Volume_2025"])/vol_compare["Volume_2025"] * 100)

vol_compare
```

```python
plu_2025=data2025[["4046", "4225", "4770"]].sum()
plu_2026=data2026[["4046", "4225", "4770"]].sum()

plu_compare = pd.DataFrame({"PLU_2025":plu_2025,"PLU_2026":plu_2026})

plu_compare["Share_2025_%"]=(plu_compare["PLU_2025"]/plu_compare["PLU_2025"].sum())*100
plu_compare["Share_2026_%"]=(plu_compare["PLU_2026"]/plu_compare["PLU_2026"].sum())*100

plu_compare
```
